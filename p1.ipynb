```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image = cv2.imread('my_photo.jpg')
image
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
height, weight = image.shape[:2]
```

```python
weight
```

```python
height
```

```python
#Convert into gray scale image
gray_image = np.zeros((height, weight), dtype= np.uint8)

for x in range(height):
    for y in range(weight):
        r,g,b = image[x,y]
        gray = int(0.299 * r + 0.587 * g + 0.114 * b)
        gray_image[x,y] = gray
        
```

```python
plt.imshow(gray_image, cmap='gray')
```

```python
#Convert image into binary image
threshold = 128
binary_image = np.zeros((height, weight), dtype= np.uint8)

for x in range(height):
    for y in range(weight):
        gray = gray_image[x,y]
        if gray<threshold:
            binary_image[x,y] = 0 #black
        else:
            binary_image[x,y] = 255 #white

#display binary image
plt.imshow(binary_image, cmap='gray')
plt.axis('off')
plt.show()
```

```python
#Display image
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("RGB Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(gray_image, cmap='gray')
plt.title("Gray scale Image")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(binary_image, cmap='gray')
plt.title("Binary scale Image")
plt.axis('off')
```
