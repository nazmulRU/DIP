```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("emma.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
smooth = np.zeros_like(image)
kernel_size = 3
```

```python
#apply averaging filter
for i in range(1,image.shape[0]-1):
    for j in range(1,image.shape[1]-1):
        
        #take 3x3 neighborhood
        neighborhood = image[i-1:i+2, j-1:j+2]
        
        #calculate average
        average = np.sum(neighborhood)/9
        
        #store result
        smooth[i,j] = int(average)
```

```python
#Display image
plt.figure(figsize=(10,6))

plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Orginal image')
plt.axis('off')


plt.subplot(1,2,2)
plt.imshow(smooth, cmap='gray')
plt.title('3 x 3 averaging filter')
plt.axis('off')

plt.tight_layout()
plt.show()
```
