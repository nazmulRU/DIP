```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('emma.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
#create output image(zero pixelled)
median = np.zeros_like(image)
maximum = np.zeros_like(image)
minimum = np.zeros_like(image)
```

```python
for i in range(1, image.shape[0]-1):
    for j in range(1, image.shape[1]-1):
        #get 3x3 neighborhood
        neighborhood = image[i-1:i+2, j-1:j+2]
        
        #median filter
        median[i,j] = np.median(neighborhood)
        
        #max filter
        maximum[i,j] = np.max(neighborhood)
        
        #min filter
        minimum[i,j] = np.min(neighborhood)
        
```

```python
#Display result
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(image, cmap='gray')
plt.title('Orginal Image')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(median, cmap='gray')
plt.title('median filter')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(maximum, cmap='gray')
plt.title('maximum filter')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(minimum, cmap='gray')
plt.title('minimum filter')
plt.axis('off')
```
