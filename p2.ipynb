```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
orginal = cv2.imread('my_photo.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(orginal,cmap='gray')
```

```python
orginal.shape
```

```python
#changing spatial resulution of the image
spatial_256 = cv2.resize(orginal,(256,256))
spatial_128 = cv2.resize(orginal,(128,128))
spatial_64 = cv2.resize(orginal,(64,64))
```

```python
#changing intensity resulution of the image
intensity_256 = (orginal//1)*1
intensity_16 = (orginal//16)*16
intensity_4 = (orginal//64)*64
```

```python
#display spatial resulution
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(orginal, cmap="gray")
plt.title('Orginal image')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(spatial_256, cmap="gray")
plt.title('256 X 256')
#plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(spatial_128, cmap="gray")
plt.title('128 X 128')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(spatial_64, cmap="gray")
plt.title('64 X 64')
plt.axis('off')
```

```python
#display intensity resulution
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(orginal, cmap="gray")
plt.title('orginal image')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(intensity_256, cmap="gray")
plt.title('256 levels')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(intensity_16, cmap="gray")
plt.title('16 levels')
#plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(intensity_4, cmap="gray")
plt.title('4 levels')
plt.axis('off')
```
